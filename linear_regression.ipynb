# Linear Regression

## What is Linear Regression?
Linear Regression is a statistical method for modeling the relationship between a scalar response (dependent variable) and one or more explanatory variables (independent variables). It attempts to model the relationship by fitting a linear equation to observed data.

## Why is it used?
1.  **Predictive Modeling**: Used to predict a continuous value (e.g., house prices, temperature) based on input features.
2.  **Trend Analysis**: Helps understand how the dependent variable changes when the independent variable(s) change.
3.  **Baseline**: Often serves as a baseline algorithm for regression tasks due to its simplicity and interpretability.

## Code Explanation

1.  **`LinearRegression` Class**:
    -   **`fit(X, y, epochs)`**:
        -   **Purpose**: Trains the model using Gradient Descent.
        -   **Significance**: Iteratively updates weights (`self.W`) and bias (`self.b`) to minimize the Mean Squared Error (MSE) between predictions (`y_hat`) and actual values (`y`). The gradients (`dW`, `db`) indicate the direction to move to reduce error.
    -   **`predict(X)`**:
        -   **Purpose**: Generates predictions for new data.
        -   **Significance**: Computes the linear application $y = XW + b$ using the learned parameters.

2.  **Usage Example**:
    -   Generates synthetic linear data with noise.
    -   Trains the model and prints the learned weights and bias.

```python
import numpy as np

class LinearRegression:
    def __init__(self, learning_rate):
        self.W = None
        self.b = None
        self.learning_rate = learning_rate

    def fit(self, X, y, epochs):
        n = X.shape[1]
        n_samples = X.shape[0]
        print(n)
        y = y.reshape(-1,1)
        self.W = np.zeros((n, 1))
        self.b = 0.0

        for _ in range(epochs):
            y_hat = X@self.W + self.b

            dW = (2/n_samples) * (X.T @ (y_hat - y))
            db = (2/n_samples) * np.sum(y_hat - y)

            self.W -= self.learning_rate*dW
            self.b -= self.learning_rate*db
    
    def   predict(self, X):
        return X@self.W + self.b

np.random.seed(0)
X = np.random.randn(100, 1)
y = 3 * X + 2 + np.random.randn(100, 1) * 0.5

model = LinearRegression(learning_rate=0.01)
model.fit(X, y, epochs=2000)

print(model.W, model.b)
```
