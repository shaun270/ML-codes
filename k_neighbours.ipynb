# K-Nearest Neighbors (KNN)

## What is KNN?
K-Nearest Neighbors is a simple, non-parametric, lazy learning algorithm used for classification and regression. Ideally, it does not "learn" a discriminative function from the training data but memorizes the training dataset instead.

## Why is it used?
1.  **Simplicity**: Easy to implement and understand. There are no assumptions about the underlying data distribution.
2.  **Versatility**: Can be used for classification, regression, and search (e.g., recommender systems).
3.  **Instance-Based**: Being a lazy learner, it adapts well when new training data is added since there is no training phase.

## Code Explanation

1.  **`k_nearest_neighbors(points, query_point, k)`**:
    -   **Purpose**: Finds the $k$ closest points in the dataset to a specific `query_point`.
    -   **Significance**:
        -   **Distance Metric**: Calculates the Euclidean distance between the `query_point` and all other `points`.
        -   **Sorting**: Sorts the points based on calculated distances (`np.argsort`).
        -   **Selection**: Selects the top $k$ points with the smallest distances. In a classification context, the majority class of these neighbors would determine the class of the query point.

```python
import numpy as np

def k_nearest_neighbors(points, query_point, k):
    points = np.array(points)
    query_point = np.array(query_point)

    
    distances = np.linalg.norm(points - query_point, axis = 1)
    x = np.argsort(distances)

    answers = []
    for i in x[:k]:
        answers.append(tuple(points[i]))

    return answers
```
