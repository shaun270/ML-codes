# K-Means Clustering

## What is K-Means Clustering?
K-Means is an iterative, unsupervised machine learning algorithm used to partition a dataset into $K$ distinct, non-overlapping subgroups (clusters). Each data point belongs to the cluster with the nearest mean (centroid), serving as a prototype of the cluster.

## Why is it used?
1.  **Exploratory Data Analysis**: Helps to discover inherent groupings in data (e.g., customer segmentation, image compression).
2.  **Simplicity & Speed**: Computationally efficient compared to hierarchical clustering for large datasets.
3.  **Versatility**: Works well when clusters are distinct and spherical.

## Code Explanation

1.  **`k_means_clustering(points, k, initial_centroids, max_iterations)`**:
    -   **Purpose**: Implements the core logic of the K-Means algorithm.
    -   **Significance**:
        -   **Distance Calculation**: Computes the Euclidean distance (`np.linalg.norm`) between every point and every centroid.
        -   **Cluster Assignment**: Assigns each point to the closest centroid (`np.argmin`).
        -   **Centroid Update**: Re-calculates the new centroid positions by taking the mean of all points currently assigned to that cluster.
        -   **Iteration**: Repeats the assignment and update steps for `max_iterations` to refine the clusters.

```python
import numpy as np

def k_means_clustering(points: list[tuple[float, float]], k: int, initial_centroids: list[tuple[float, float]], max_iterations: int) -> list[tuple[float, float]]:
    cluster_points = np.array(initial_centroids)
    points = np.array(points)
    for j in range(max_iterations):
        indexes = []
        for i in range(points.shape[0]):
            distances = np.linalg.norm(cluster_points - points[i], axis = 1)
            indexes.append(np.argmin(distances))
        for i in range(k):
            boolean_array = [ x == i for x in indexes]
            cluster = points[boolean_array]
            cluster_points[i] = np.mean(cluster, axis = 0)
   
    rounded = np.round(cluster_points, 4)
    final_centroids = [tuple(i) for i in rounded]
    return final_centroids
```
